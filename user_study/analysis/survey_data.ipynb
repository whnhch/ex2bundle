```python
post_study_qs = []
post_sbert = []
post_sudocu = []

for u_id in user_ids:

    df_post_study = pd.DataFrame.from_dict(data[u_id]["postStudy"])
    df_post_sbert = pd.DataFrame.from_dict(data[u_id]["postTaskSbert"])
    df_post_sudocu = pd.DataFrame.from_dict(data[u_id]["postTaskSudocu"])

    post_study_qs.append(df_post_study.transpose()["answer"].to_numpy())
    post_sbert.append(df_post_sbert.transpose()["answer"].to_numpy())
    post_sudocu.append(df_post_sudocu.transpose()["answer"].to_numpy())


print(post_study_qs[0])
```

```python
num_sudocu = 0
num_sbert = 0

idx_of_interest = 8

for u_ans in post_study_qs:

    print(u_ans[idx_of_interest])

    if u_ans[idx_of_interest] == "System U":
        num_sudocu += 1
    else:
        num_sbert += 1

print("num_sudocu: %d, num_sbert: %d" % (num_sudocu, num_sbert))
```
