```python
import json
import numpy as np
import pandas as pd
from nltk import sent_tokenize, word_tokenize
import random
```

```python
top_words = None

with open("topic_words_json.txt") as f:
    top_words = json.load(f)

top_words = top_words['words']

print(len(top_words[0]))
```

```python
words = ["north", "eastern", "west", "east", "south", "mountains", "river"]

for id, topic in enumerate(top_words):

    ctr = 0

    for word in words:

        if word in topic:
            ctr += 1

    print("word match count: %d for topic %d" % (ctr, id))
```
