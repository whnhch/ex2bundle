```python
import scipy.stats as st

rouge_cols = ["user_id", "r1f1", "r2f1", "r3f1", "relax_type"]

r_f1_df = df[rouge_cols]

mean_all = np.mean(df[["r1f1", "r2f1", "r3f1"]].to_numpy(), axis=0)
cis_ = st.norm.interval(alpha=0.95, loc=mean_all, scale=st.sem(df[["r1f1", "r2f1", "r3f1"]].to_numpy(), axis=0))
errors_all = (cis_[1] - cis_[0]) / 2

mean_doc = np.mean(df[df["relax_type"] == "Doc"][["r1f1", "r2f1", "r3f1"]].to_numpy(),axis=0)
cis_ = st.norm.interval(alpha=0.95, loc=mean_doc, scale=st.sem(df[df["relax_type"] == "Doc"][["r1f1", "r2f1", "r3f1"]].to_numpy(), axis=0))
errors_doc = (cis_[1] - cis_[0]) / 2

mean_re = np.mean(df[df["relax_type"] == "Slider"][["r1f1", "r2f1", "r3f1"]].to_numpy(),axis=0)
cis_ = st.norm.interval(alpha=0.95, loc=mean_re, scale=st.sem(df[df["relax_type"] == "Slider"][["r1f1", "r2f1", "r3f1"]].to_numpy(), axis=0))
errors_re = (cis_[1] - cis_[0]) / 2



sbert_mean = np.mean(df_sbert[["r1f1", "r2f1", "r3f1"]].to_numpy(), axis=0)
cis_ = st.norm.interval(alpha=0.95, loc=sbert_mean, scale=st.sem(df_sbert[["r1f1", "r2f1", "r3f1"]].to_numpy(), axis=0))
sbert_errors = (cis_[1] - cis_[0]) / 2
```

```python
# user_ids = ['NIWAYe4A', '2OaKfVJy', '2P6OXwHJ']

sudocu_user_data = data_parser.processed_user_interactions["sudocu"]

user_sudocu_system_runtimes = None

with open("sudocu_system_runtimes_time_analysis.json", "r") as f:
   user_sudocu_system_runtimes = json.load(f)
   f.close()


for u_id in user_ids:
    print(u_id)
    print("Min Runtime: %.4f" % np.min(user_sudocu_system_runtimes[u_id]))
    print("Max Runtime: %.4f" % np.max(user_sudocu_system_runtimes[u_id]))

u_id_avgs = {}

df_lines = []

df_column_names = ["u_id", "avg_avg_cplex_runtime", "avg_total_cplex_runtime", "avg_total_runtime", "avg_cplex_total_ratio", "avg_avg_buildtime",
                    "avg_total_buildtime", "avg_build_ratio", "avg_avg_loadtime", "avg_total_loadtime", "avg_load_ratio", 
                    "avg_total_sudocu", "avg_sudocu_ratio", "avg_cplex_iters", "avg_problem_size", "avg_avg_relax_time", "avg_total_relax_time", "avg_relax_time_ratio"]

for u_id in user_ids:

    print(u_id)

    num_generated = len(user_sudocu_system_runtimes[u_id])

    print(num_generated)
    print(len(sudocu_user_data[u_id]["generated_summaries"]["summary_stats"]))
    
    u_id_avgs[u_id] = {"num_gen":num_generated}

    avg_cplex_time = []
    avg_build_time = []
    total_cplex_time = []
    cplex_total_ratio = []
    num_cplex_iters = []
    problem_size = []
    build_time = []
    build_time_ratio = []

    avg_load_time = []
    total_load_time = []
    load_time_ratio = []

    avg_relax_time = []
    total_relax_time = []
    relax_time_ratio = []

    total_sudocu_time = []
    sudocu_time_ratio = []

    for i, (total_runtime, feedback_stats) in enumerate(zip(user_sudocu_system_runtimes[u_id], sudocu_user_data[u_id]["generated_summaries"]["summary_stats"])):
        _avg_cplex_time = np.mean(feedback_stats["model_feedback"]["cplex_runtimes"])
        _total_cplex_time = np.sum(feedback_stats["model_feedback"]["cplex_runtimes"])
        _cplex_total_ratio = _total_cplex_time / total_runtime
        
        _num_cplex_iter = feedback_stats["model_feedback"]["solve_iter"]
        _problem_size = feedback_stats["model_feedback"]["problem_size"]
        
        _avg_build_time = np.mean(feedback_stats["model_feedback"]["build_time"])
        _total_build_time = np.sum(feedback_stats["model_feedback"]["build_time"])
        _build_time_ratio = _total_build_time / total_runtime

        _avg_load_time = np.mean(feedback_stats["model_feedback"]["load_time"])
        _total_load_time = np.sum(feedback_stats["model_feedback"]["load_time"])
        _load_time_ratio = _total_load_time / total_runtime

        _avg_relax_time = np.mean(feedback_stats["model_feedback"]["relax_time"])
        _total_relax_time = np.sum(feedback_stats["model_feedback"]["relax_time"])
        _relax_time_ratio = _total_relax_time / total_runtime

        _total_measured_sudocu_time = _total_build_time + _total_cplex_time + _total_load_time + _total_relax_time
        _total_measured_ratio = _total_measured_sudocu_time / total_runtime

        avg_cplex_time.append(_avg_cplex_time)
        total_cplex_time.append(_total_cplex_time)
        cplex_total_ratio.append(_cplex_total_ratio)
        num_cplex_iters.append(_num_cplex_iter)
        problem_size.append(_problem_size)
        build_time.append(_total_build_time)
        build_time_ratio.append(_build_time_ratio)
        avg_build_time.append(_avg_build_time)
        avg_load_time.append(_avg_load_time)
        total_load_time.append(_total_load_time)
        load_time_ratio.append(_load_time_ratio)
        total_sudocu_time.append(_total_measured_sudocu_time)
        sudocu_time_ratio.append(_total_measured_ratio)
        avg_relax_time.append(_avg_relax_time)
        total_relax_time.append(_total_relax_time)
        relax_time_ratio.append(_relax_time_ratio)

    # u_id_avgs[u_id]["avg_avg_cplex_runtime"] = np.mean(avg_cplex_time)
    # u_id_avgs[u_id]["avg_total_cplex_runtime"] = np.mean(total_cplex_time)
    # u_id_avgs[u_id]["avg_total_runtime"] = np.mean(user_sudocu_system_runtimes[u_id])
    # u_id_avgs[u_id]["avg_cplex_total_ratio"] = np.mean(cplex_total_ratio)
    # u_id_avgs[u_id]["avg_cplex_iteration"] = np.mean(num_cplex_iters)
    # u_id_avgs[u_id]["avg_problem_size"] = np.mean(problem_size)
    # u_id_avgs[u_id]["avg_avg_build_time"] = np.mean(build_time)
    # u_id_avgs[u_id]["avg_buildtime_ratio"] = np.mean(build_time_ratio)

    # u_id_avgs[u_id]["avg_avg_cplex_runtimes"] = avg_cplex_time
    # u_id_avgs[u_id]["avg_total_cplex_runtimes"] = total_cplex_time
    # u_id_avgs[u_id]["avg_total_runtimes"] = user_sudocu_system_runtimes[u_id]
    # u_id_avgs[u_id]["avg_cplex_total_ratios"] = cplex_total_ratio
    # u_id_avgs[u_id]["avg_cplex_iterations"] = num_cplex_iters
    # u_id_avgs[u_id]["problem_sizes"] = problem_size
    # u_id_avgs[u_id]["build_times"] = build_time
    # u_id_avgs[u_id]["buildtime_ratios"] = build_time_ratio

    df_line = [u_id, round(np.mean(avg_cplex_time), 4), round(np.mean(total_cplex_time), 4), round(np.mean(user_sudocu_system_runtimes[u_id]), 4),
                round(np.mean(cplex_total_ratio), 4), round(np.mean(avg_build_time), 4), round(np.mean(build_time), 4), round(np.mean(build_time_ratio), 4),
                round(np.mean(avg_load_time), 4), round(np.mean(total_load_time), 4), round(np.mean(load_time_ratio), 4),
                round(np.mean(total_sudocu_time), 4), round(np.mean(sudocu_time_ratio), 4),
                round(np.mean(num_cplex_iters), 2), round(np.mean(problem_size), 2), round(np.mean(avg_relax_time), 4),
                round(np.mean(total_relax_time), 4), round(np.mean(relax_time_ratio), 4)]

    df_lines.append(df_line)

    print("Avg - avg CPLEX runtimes: ", round(np.mean(avg_cplex_time), 4))
    print("Avg - total CPLEX runtimes: ", round(np.mean(total_cplex_time), 4))
    print("Avg - total runtimes: ", round(np.mean(user_sudocu_system_runtimes[u_id]), 4))
    print("Avg - CPLEX/total runtime ratios: ", round(np.mean(cplex_total_ratio), 4))

    print("Avg - avg CPLEX buildtimes: ", round(np.mean(avg_build_time), 4))
    print("Avg - total CPLEX buildtimes: ", round(np.mean(build_time), 4))
    print("Avg - CPLEX/total buildtime/runtime ratios: ", round(np.mean(build_time_ratio), 4))

    print("Avg - avg CPLEX loadtimes: ", round(np.mean(avg_load_time), 4))
    print("Avg - total CPLEX loadtimes: ", round(np.mean(total_load_time), 4))
    print("Avg - CPLEX/total loadtime/runtime ratios: ", round(np.mean(load_time_ratio), 4))

    print("Avg - avg CPLEX relax-times: ", round(np.mean(avg_relax_time), 4))
    print("Avg - total CPLEX relax-times: ", round(np.mean(total_relax_time), 4))
    print("Avg - CPLEX/total relax-time/runtime ratios: ", round(np.mean(relax_time_ratio), 4))

    print("Avg - total SuDocu time: ", round(np.mean(total_sudocu_time), 4))
    print("Avg - SuDocu/total runtime ratios: ", round(np.mean(sudocu_time_ratio), 4))

    print("Avg - CPLEX iterations: ", round(np.mean(num_cplex_iters), 2))
    print("Avg - CPLEX Problem Size: ", round(np.mean(problem_size), 2), "\n")


sudocu_runtimes_df = pd.DataFrame(data=df_lines, columns=df_column_names)
```
