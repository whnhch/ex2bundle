```python
lines = []

with open("complete_u_study_02.txt", "r") as f:
    lines = f.readlines()
    f.close()

print(len(lines))
```

```python
user_runtime_dict_sudocu = {}

all_runtimes = []

for u_id in user_ids:

    user_runtime_dict_sudocu[u_id] = []

    for idx, line in enumerate(lines):
        # if "{:s}Generating a SuDocu Summary!!!!".format(u_id) in line:

        #     # get the runtime from 3 lines back
        #     val = float(lines[idx-3].split("time: ")[1].split(" ")[1].strip())

        #     user_runtime_dict_sudocu[u_id].append(val)

        if "{:s}  - Subprocess execution time:".format(u_id) in line:


            for i in range(1, 5):

                if "SystemU" in lines[idx-i]:

                    val = float(line.split("time: ")[1].strip())

                    user_runtime_dict_sudocu[u_id].append(val)

                    all_runtimes.append(val)

                    break
            
            # print(lines[idx-1])
            # print(lines[idx-2])
            # print(lines[idx-3])
            # print(lines[idx-4])
```

```python
print(np.mean(all_runtimes))
print(np.std(all_runtimes))

print(min(all_runtimes))
print(max(all_runtimes))
```
